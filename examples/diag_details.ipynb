# Diagnostic Details

This notebook shows how to inspect cached LHD diagnostics with the generic xarray description helpers. It is meant for checking coordinate names, dimensions, signal names, shapes, and units before plotting or analysis.


## Setup

Use the existing local cache and load diagnostics without triggering downloads.


```python
from pathlib import Path

from lhd_data.describe import describe_dataarray, describe_dataset, describe_many
from lhd_data.plotting.helpers import (
    DEFAULT_OVERVIEW_DIAGNOSTICS,
    load_cached_diagnostics,
    summarize_datasets,
)

SHOT = 193819
CACHE_DIR = Path("../local/bh-molecule")
DIAGNOSTICS = DEFAULT_OVERVIEW_DIAGNOSTICS
```

## Load And Summarize

`load_cached_diagnostics` returns loaded datasets and a mapping of failures. The summary stays compact enough for quick notebook use.


```python
datasets, failed = load_cached_diagnostics(SHOT, DIAGNOSTICS, cache_dir=CACHE_DIR)
print(summarize_datasets(datasets, failed, diagnostics=DIAGNOSTICS))
```

## One Dataset

Use `describe_dataset` when you want all coordinates, dimensions, and signal names for one diagnostic.


```python
print(describe_dataset(datasets["thomson"]))
```

## Several Related Datasets

Use `describe_many` for grouped diagnostics such as `ha1` and `ha2`.


```python
halpha = {name: datasets[name] for name in ("ha1", "ha2") if name in datasets}
print(describe_many(halpha))
```

## One Signal

Use `describe_dataarray` when you already know the variable you care about and want a one-line check. The overview plot uses `ha2` `3-O(H)` as the default H-alpha signal.


```python
print(describe_dataarray(datasets["ha2"]["3-O(H)"]))
```
